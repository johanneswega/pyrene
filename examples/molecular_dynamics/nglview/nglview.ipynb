# Visulize Simulation boxes and trajectories inside Jupyter Notebooks

Using the `nglview` and `MDAnalysis` python packages it is quite straight forward to view simulation boxes (`.gro` files) or whole MD trajectories (`.xtc`) inside Jupyter Notebooks. 

For this make sure that you are in a dedicated conda environment. For me: `conda activate nglview`. 

One note: I had to install `python -m pip install "nglview==3.1.4"` for it work, with the newest version of nglview I got some errors. 

It visualize trajectories it is then quite straightforward:

(If you just want to visulize `.gro` files, remove the second entry in the mda.universe command)

```python
import MDAnalysis as mda
import nglview as nv
from IPython.display import display

u1 = mda.Universe(
    "BPe_DPA_IP_in_DCM/prod/prod.gro",
    "BPe_DPA_IP_in_DCM/prod/prod-mol.xtc"
)
```

```python
view = nv.show_mdanalysis(u1)

view.clear_representations()

view.frame = 1

view.add_representation(
    "licorice",
    selection="BPE",
)

view.add_representation(
    "licorice",
    selection="DPA",
)

view.add_representation(
    "line",
    selection="DCM",
    opacity=0.2
)

view
```

To export a high quality image, drag the viewer to the appropriate frame and position it as wanted and run the following code: 

(increase factor for higher quality)


```python
view.download_image(
    filename="test.png",
    factor=4,
    antialias=True,
    trim=True,
    transparent=False
)
```

# Visualizing Molecular Strucutres

Below is just a short code snipit used to make a .png of a molecular .gro file. Neat to compile a list of the different molecules within the simulation. 

```python
u = mda.Universe("BPe_DPA_IP_in_ACN/ACN.gro")
view = nv.show_mdanalysis(u)
view.clear_representations()
view.add_representation("licorice",selection="ACN")
view
```

```python
view.download_image(filename="ACN.png",factor=4,antialias=True,trim=True,transparent=False)
```
